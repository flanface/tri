# tri: learning notebooks

tri is a backend learning engine for Deepwoken talent and mantra selection. It analyses a build, completes a partial build and learns from expert reviews. These notebooks replace the written docs; each one runs on its own, and the cells marked **needs a run** use a fitted run and a snapshot of the review tables.

| notebook | what it covers | needs a run |
|---|---|---|
| `01_api` | the public module `tri.api`: loading, reviewing, completing, dry inserts | yes |
| `02_similarity_and_chasm` | build similarity $\sigma$, the chasm factor $\chi$ and supersession $f(\sigma)$ | no |
| `03_reach_and_the_logit` | the directed reach $K_c$, the per-item logit $z_t$ and slot filling | no |
| `04_evaluation` | leave-one-build-out $F_1$, review agreement, bootstrap errors and every measured number | optional |
| `05_bot_setup` | the Discord bot, its environment, runs, hosting and plots on GitHub Pages | no |
